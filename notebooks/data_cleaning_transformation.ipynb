```python
import pandas as pd
```

```python
file_path = r"D:\nava-ai\data\raw\online_retail_II.xlsx"
df = pd.read_excel(file_path)
df_clean = df.copy()
```

```python
print("Original Data SHape:",df.shape)
print("Clean Data Shape:",df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
)
df_clean.columns
```

```python
df_clean.info()
```

```python
df_clean["invoicedate"] = pd.to_datetime(
    df_clean["invoicedate"],
    errors="coerce"
)
#df_clean["invoicedate"].head()
df_clean["quantity"] = pd.to_numeric(
    df_clean["quantity"],
    errors="coerce"
)
df_clean["price"] = pd.to_numeric(
    df_clean["quantity"],
    errors="coerce"
)
df_clean["customer_id"] = pd.to_numeric(
    df_clean["quantity"],
    errors="coerce"
)
```

```python
df_clean.info()
```

```python
duplicate_count=df_clean.duplicated().sum()
print("Duplicate count:",duplicate_count)
duplicate_percentage = (
    duplicate_count / len(df_clean) * 100
)

print(
    f"Duplicate percentage: {duplicate_percentage:.2f}%"
)
```

```python
duplicate_rows = df_clean[
    df_clean.duplicated(keep=False)
].sort_values(
    by=["invoice", "stockcode", "invoicedate"]
)

duplicate_rows.head(20)
```

```python
duplicate_groups = (
    df_clean
    .groupby(
        [
            "invoice",
            "stockcode",
            "description",
            "quantity",
            "invoicedate",
            "price",
            "customer_id",
            "country"
        ],
        dropna=False
    )
    .size()
)

duplicate_groups[
    duplicate_groups > 1
].sort_values(
    ascending=False
).head(20)
```

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before:", before)
print("Rows after :", after)
print("Rows removed:", before - after)
print("Remaining duplicates:", df_clean.duplicated().sum())
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": df_clean.isnull().sum(),
    "Missing %": (
        df_clean.isna().sum() / len(df_clean) * 100
    )
})

missing_summary
```

```python
print("Missing Customer IDs:",
      df_clean["customer_id"].isnull().sum())

print("Missing Descriptions:",
      df_clean["description"].isnull().sum())
```

```python
missing_description = df_clean[
    df_clean["description"].isna()
]

print("Rows with missing Description:",
      len(missing_description))
```

```python
missing_description[
    [
        "invoice",
        "stockcode",
        "quantity",
        "invoicedate",
        "price",
        "customer_id",
        "country"
    ]
].head(30)
```

```python
missing_desc_codes = df_clean.loc[
    df_clean["description"].isna(),
    "stockcode"
].unique()

print("Unique StockCodes with missing Description:",
      len(missing_desc_codes))

missing_desc_codes[:20]
```

```python
description_lookup = (
    df_clean.groupby("stockcode")["description"].first()
)

df_clean["description"] = (
    df_clean["description"]
    .fillna(df_clean["stockcode"].map(description_lookup))
)
```

```python
print("Missing descriptions:",
      df_clean["description"].isna().sum())
```
