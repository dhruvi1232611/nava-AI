```python
import pandas as pd
import numpy as np
```

```python
file_path = r"D:\nava-ai\data\raw\online_retail_II.xlsx"
df = pd.read_excel(file_path)
df_clean = df.copy()
```

```python
print("Original Data SHape:",df.shape)
print("Clean Data Shape:",df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
)
df_clean.columns
```

```python
df_clean["invoicedate"] = pd.to_datetime(
    df_clean["invoicedate"],
    errors="coerce"
)
#df_clean["invoicedate"].head()
df_clean["quantity"] = pd.to_numeric(
    df_clean["quantity"],
    errors="coerce"
)
df_clean["price"] = pd.to_numeric(
    df_clean["price"],
    errors="coerce"
)
df_clean["customer_id"] = pd.to_numeric(
    df_clean["customer_id"],
    errors="coerce"
)
```

```python
df_clean["price_status"] = np.where(
    df_clean["price"] < 0,
    "Invalid/Adjustment Price",
    "Valid Price"
)
df_clean.tail(30)
```

**Revenue Generation**

**---> Sales Revenue**

```python
df_clean["sales_revenue"] = np.where(
    (df_clean["quantity"] > 0) & (df_clean["price"] > 0),
    df_clean["quantity"] * df_clean["price"],
    0
)
df_clean.head(10)
```

**---> Return Revenue**

```python
df_clean["return_amount"] = np.where(
    (df_clean["quantity"] < 0) & (df_clean["price"] > 0),
    abs(df_clean["quantity"] * df_clean["price"]),
    0
)
df_clean.tail()
```

**---> Net Revenue**

```python
df_clean["net_revenue"] = (
    df_clean["sales_revenue"] - df_clean["return_amount"]
)
df_clean.head()
```

**Date and time features generation**

```python
df_clean["date"] = df_clean["invoicedate"].dt.date
```

```python
df_clean["year"] = df_clean["invoicedate"].dt.year
```

```python
df_clean["month"] = df_clean["invoicedate"].dt.month
```

```python
df_clean["month_name"] = df_clean["invoicedate"].dt.month_name()
```

```python
df_clean["day"] = df_clean["invoicedate"].dt.day
```

```python
df_clean["day_of_week"] = df_clean["invoicedate"].dt.day_name()
```

```python
df_clean["hour"] = df_clean["invoicedate"].dt.hour
```

```python
df_clean["is_weekend"] = (
    df_clean["invoicedate"].dt.dayofweek >= 5
)
```
