**Customer Level Features**

```python
import pandas as pd
import numpy as np
```

```python
file_path = r"D:\nava-ai\data\clean\retail_buisness_cleaned_data.xlsx"
df_clean = pd.read_excel(file_path)
df_clean.head(10)
```

```python
customer_features = df_clean.groupby("customer_id").agg(
    total_orders=("invoice", "nunique")
).reset_index()
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["quantity"].sum().reset_index(
        name= "total_quantity"
    ),
    on="customer_id",
    how="left"
)
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["sales_revenue"].sum().reset_index(
        name= "total_sales"
    ),
    on="customer_id",
    how="left"
)
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["return_amount"].sum().reset_index(
        name= "total_return_amount"
    ),
    on="customer_id",
    how="left"
)
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["net_revenue"].sum().reset_index(
        name= "net_revenue"
    ),
    on="customer_id",
    how="left"
)
```

```python
customer_features.head()
```

```python
customer_features["avg_oreder_value"] = (
    customer_features["total_sales"] /
    customer_features["total_orders"]
)
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["invoicedate"].min().reset_index(
        name="first_purchase_date"
    )
    ,
    on="customer_id",
    how="left"
)
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["invoicedate"].max().reset_index(
        name = "last_purchase_date"
    ),
    on="customer_id",
    how="left"
)
```

```python
customer_features.head()
```

```python
customer_features = customer_features.merge(
    df_clean.groupby("customer_id")["stockcode"].nunique().reset_index(
        name = "unique_product_purchased"
    ),
    on = "customer_id",
    how = "left"
)
```

```python
customer_features.head()
```

```python
customer_features["number_of_active_days"] = (customer_features["last_purchase_date"] - customer_features["first_purchase_date"]).dt.days
```

```python
customer_features.head()
```

```python
reference_date = df_clean["invoicedate"].max()
```

```python
customer_features["recency_days"] = (reference_date - customer_features["last_purchase_date"]).dt.days
```

```python
customer_features[["customer_id","last_purchase_date","recency_days"]].head()
```

```python
customer_features.to_csv(
    "customer_features.csv",
    index=False
)
```

```python
customer_features["frequency_score"] = pd.qcut(
    customer_features["total_orders"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
customer_features[["customer_id" , "total_orders" , "frequency_score"]]
```

```python
customer_features["recency_score"] = pd.qcut(
    customer_features["recency_days"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)
```

```python
customer_features[
    ["customer_id", "recency_days", "recency_score"]
].head(10)
```

```python
customer_features["monetary_score"] = pd.qcut(
    customer_features["net_revenue"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
customer_features[
    ["customer_id", "net_revenue", "monetary_score"]
].head(10)
```

```python
customer_features["rfm_score"] = (
    customer_features["recency_score"].astype(str) +
    customer_features["frequency_score"].astype(str) +
    customer_features["monetary_score"].astype(str)
)
```

```python
customer_features[
    [
        "customer_id",
        "recency_score",
        "frequency_score",
        "monetary_score",
        "rfm_score"
    ]
].head(10)
```

```python
customer_features.to_csv(
    r"D:\nava-ai\data\features\customer_features.csv",
    index=False
)
```

```python
customer_features.columns.to_list()
```

**Product Level Features**

```python
product_features = df_clean.groupby("stockcode").agg(
    total_orders = ("invoice","nunique")
).reset_index()
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean[df_clean["quantity"] > 0]
    .groupby("stockcode")["quantity"]
    .sum()
    .reset_index(name="total_units_sold"),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean[df_clean["quantity"] < 0]
    .groupby("stockcode")["quantity"]
    .sum()
    .abs()
    .reset_index(name="total_units_returned"),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean.groupby("stockcode")["sales_revenue"].sum().reset_index(
        name="total_sales"
    ),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean.groupby("stockcode")["return_amount"].sum().reset_index(
        name="total_return_amount"
    ),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean.groupby("stockcode")["net_revenue"].sum().reset_index(
        name="total_net_revenue"
    ),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean.groupby("stockcode")["customer_id"].nunique().reset_index(
        name="unique_customer"
    ),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean[
        (df_clean["quantity"] > 0) &
        (df_clean["price"] > 0)
    ]
    .groupby("stockcode")["price"]
    .mean()
    .reset_index(name="average_price"),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features = product_features.merge(
    df_clean[df_clean["quantity"] > 0]
    .groupby("stockcode")["date"]
    .nunique()
    .reset_index(name="active_selling_days"),
    on="stockcode",
    how="left"
)
```

```python
product_features.head()
```

```python
product_features.to_csv(
    r"D:\nava-ai\data\features/product_features.csv",
    index=False
)
```

**Time Level features**

```python
time_features = df_clean.groupby("date").agg(
    total_orders = ("invoice" , "nunique")
).reset_index()
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean[df_clean["quantity"] > 0]
    .groupby("date")["quantity"]
    .sum()
    .reset_index(name="total_units_sold"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean[df_clean["quantity"] < 0]
    .groupby("date")["quantity"]
    .sum()
    .abs()
    .reset_index(name="total_units_returned"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean.groupby("date")["sales_revenue"]
    .sum()
    .reset_index(name="total_sales"),
    on="date",
    how="left"
    
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean.groupby("date")["return_amount"]
    .sum()
    .reset_index(name="total_return_amount"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean.groupby("date")["net_revenue"]
    .sum()
    .reset_index(name="net_revenue"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean.groupby("date")["customer_id"]
    .nunique()
    .reset_index(name="unique_customers"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.merge(
    df_clean[df_clean["quantity"] > 0]
    .groupby("date")["stockcode"]
    .nunique()
    .reset_index(name="unique_products_sold"),
    on="date",
    how="left"
)
```

```python
time_features.head()
```

```python
time_features = time_features.drop(
    columns=["total_units_returned_x"]
)
```

```python
time_features.head()
```

```python
time_features.to_csv(
    r"D:\nava-ai\data\features\time_features.csv",
    index=False
)
```
