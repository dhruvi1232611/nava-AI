**Monthly revenue**

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
monthly_revenue = (
    df.set_index("InvoiceDate")
      .resample("ME")["Revenue"]
      .sum()
)

monthly_revenue
```

```python
plt.figure(figsize=(12, 5))

monthly_revenue.plot()

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Monthly Net Quantity**

```python
monthly_quantity = (
    df.set_index("InvoiceDate")
      .resample("ME")["Quantity"]
      .sum()
)

monthly_quantity.plot(figsize=(12, 5))

plt.title("Monthly Quantity Sold")
plt.xlabel("Month")
plt.ylabel("Quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Actual Sold Units**

```python
sales_df = df[df["Quantity"] > 0]

monthly_quantity_sold = (
    sales_df.set_index("InvoiceDate")
              .resample("ME")["Quantity"]
              .sum()
)

monthly_quantity_sold.plot(figsize=(12, 5))

plt.title("Monthly Units Sold")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Separate positive and negative quantity transactions
sales_df = df[df["Quantity"] > 0].copy()
negative_df = df[df["Quantity"] < 0].copy()

# Monthly units sold
monthly_sold = (
    sales_df
    .set_index("InvoiceDate")["Quantity"]
    .resample("ME")
    .sum()
)

# Monthly returned/cancelled units
monthly_returns = (
    negative_df
    .set_index("InvoiceDate")["Quantity"]
    .resample("ME")
    .sum()
    .abs()
)


# Combine into one table
monthly_summary = pd.DataFrame({
    "Units Sold": monthly_sold,
    "Returned / Cancelled Units": monthly_returns
})


# Fill missing months with 0
monthly_summary = monthly_summary.fillna(0)

# Calculate net units
monthly_summary["Net Units"] = (
    monthly_summary["Units Sold"]
    - monthly_summary["Returned / Cancelled Units"]
)

monthly_summary["Return Rate (%)"] = (
    monthly_summary["Returned / Cancelled Units"]
    / monthly_summary["Units Sold"]
    * 100
)

monthly_summary
```

```python
# Make a copy for visualization
plot_df = monthly_summary.copy()

# Convert index to readable month labels
plot_df.index = plot_df.index.strftime("%b %Y")

# Create figure
fig, ax = plt.subplots(figsize=(16, 7))

# Bar positions
x = np.arange(len(plot_df))
width = 0.25

# Create grouped bars
bars1 = ax.bar(
    x - width,
    plot_df["Units Sold"],
    width,
    label="Units Sold"
)

bars2 = ax.bar(
    x,
    plot_df["Returned / Cancelled Units"],
    width,
    label="Returned / Cancelled"
)

bars3 = ax.bar(
    x + width,
    plot_df["Net Units"],
    width,
    label="Net Units"
)

# Titles and labels
ax.set_title(
    "NAVA | Monthly Operations Overview",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Units", fontsize=12)

# Show EVERY month
ax.set_xticks(x)
ax.set_xticklabels(
    plot_df.index,
    rotation=45,
    ha="right"
)

# Add grid
ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# Add legend
ax.legend(
    loc="upper left"
)

# Format Y-axis numbers with commas
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda value, _: f"{int(value):,}")
)

plt.tight_layout()
plt.show()
```

**relationship betwwen quantity and price**

```python
sample_df = df[["Quantity", "Price"]].sample(
    n=min(10000, len(df)),
    random_state=42
)

plt.figure(figsize=(10, 6))

snb.scatterplot(
    data=sample_df,
    x="Price",
    y="Quantity",
    alpha=0.4
)

plt.title("Quantity vs Price")
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.tight_layout()
plt.show()
```

```python
# Select numerical features for correlation analysis
numeric_features = df[
    ["Quantity", "Price", "Revenue"]
]

# Calculate correlation matrix
correlation_matrix = numeric_features.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))

snb.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("NAVA | Correlation Heatmap")
plt.tight_layout()
plt.show()
```
